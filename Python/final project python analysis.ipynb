```python

import pandas as pd
df = pd.read_csv(r"C:\Users\Radwa\Downloads\archive (1)\medical_insurance.csv")
df.head()
```

```python
# Count and print the total number of rows in the dataset
total_rows = len(df)
print(total_rows)
```

```python
# Display the first 10 rows to get a quick preview of the data
print(df.head(10))
```

```python
# Print the names of all columns in the dataset
print(df.columns)
```

```python
# Calculate key summary statistics: min/max/avg medical cost, avg BMI, and age range
summary = {
    "min_cost": df["annual_medical_cost"].min(),
    "max_cost": df["annual_medical_cost"].max(),
    "avg_cost": round(df["annual_medical_cost"].mean(), 2),
    "avg_bmi": round(df["bmi"].mean(), 2),
    "min_age": df["age"].min(),
    "max_age": df["age"].max()
}

print(summary)
```

```python
# Group by smoking status and calculate count, average cost, and max cost per group
smoker_analysis = df.groupby("smoker").agg(
    total_people=("smoker", "count"),
    avg_cost=("annual_medical_cost", "mean"),
    max_cost=("annual_medical_cost", "max")
).round(2)

print(smoker_analysis)
```

```python
# Group by region and compute count, average cost, and average BMI — sorted by highest cost
region_analysis = df.groupby("region").agg(
    total_people=("region", "count"),
    avg_cost=("annual_medical_cost", "mean"),
    avg_bmi=("bmi", "mean")
).round(2).sort_values("avg_cost", ascending=False)

print(region_analysis)
```

```python
# Group by sex to compare total count, average cost, and average BMI between genders
sex_analysis = df.groupby("sex").agg(
    total=("sex", "count"),
    avg_cost=("annual_medical_cost", "mean"),
    avg_bmi=("bmi", "mean")
).round(2)

print(sex_analysis)
```

```python
# Group by number of chronic conditions and compute average cost per group
chronic_analysis = df.groupby("chronic_count").agg(
    total_people=("chronic_count", "count"),
    avg_cost=("annual_medical_cost", "mean")
).round(2)

print(chronic_analysis)
```

```python
# Group by BMI category and compute total count and average medical cost per category
bmi_analysis = df.groupby(
    "bmi_category",
    observed=True
).agg(
    total=("bmi_category", "count"),
    avg_cost=("annual_medical_cost", "mean")
).round(2)

print(bmi_analysis)
```

```python
# Classify each person into an age group, then compute count and average cost per group
def age_group(age):
    if age < 30:
        return "Young"
    elif age < 45:
        return "Middle"
    elif age < 60:
        return "Senior"
    else:
        return "Elder"

df["age_group"] = df["age"].apply(age_group)

age_analysis = df.groupby("age_group").agg(
    total=("age_group", "count"),
    avg_cost=("annual_medical_cost", "mean")
).round(2)

print(age_analysis)
```

```python
# Cross-analyze smoking status and BMI category together to see combined effect on cost
combo = df.groupby(["smoker", "bmi_category"]).agg(
    total=("smoker", "count"),
    avg_cost=("annual_medical_cost", "mean")
).round(2)

print(combo)
```

```python
# Bucket each person's annual cost into predefined ranges and count how many fall in each
bins = [0, 5000, 10000, 20000, 50000, float("inf")]
labels = ["0-5K", "5K-10K", "10K-20K", "20K-50K", "50K+"]

df["cost_range"] = pd.cut(
    df["annual_medical_cost"],
    bins=bins,
    labels=labels
)

print(df["cost_range"].value_counts().sort_index())
```

```python
# Filter only above-average-cost individuals, then group by smoking, sex, and age group
avg_cost = df["annual_medical_cost"].mean()

high_cost = df[df["annual_medical_cost"] > avg_cost]

result = high_cost.groupby(
    ["smoker", "sex", "age_group"]
).size().reset_index(name="high_cost_cases")

print(result)
```

```python
# Rank each person's cost within their region
df["rank_in_region"] = df.groupby("region")[
    "annual_medical_cost"
].rank(method="dense", ascending=False)

top_5_in_region = df[df["rank_in_region"] <= 5]

top_5_in_region[[
    "person_id",
    "region",
    "annual_medical_cost",
    "rank_in_region"
]]
```

```python
# Rank each person's cost within their region (1 = most expensive in that region)
df["region_avg"] = df.groupby("region")[
    "annual_medical_cost"
].transform("mean")

df["diff_from_avg"] = (
    df["annual_medical_cost"] - df["region_avg"]
)

print(
    df[[
        "person_id",
        "region",
        "annual_medical_cost",
        "region_avg",
        "diff_from_avg"
    ]].head(5)
)
```

```python
# Compute the full statistical summary of annual medical cost (min, max, avg, Q1, median, Q3)
stats = {
    "min": round(df["annual_medical_cost"].min(), 2),
    "max": round(df["annual_medical_cost"].max(), 2),
    "avg": round(df["annual_medical_cost"].mean(), 2),
    "Q1": round(df["annual_medical_cost"].quantile(0.25), 2),
    "Median": round(df["annual_medical_cost"].quantile(0.50), 2),
    "Q3": round(df["annual_medical_cost"].quantile(0.75), 2)
}

print(stats)
```

```python
# For each region, count how many people exceed $30K and what percentage that represents
high_cost_region = df.groupby("region").apply(
    lambda x: pd.Series({
        "total_people": len(x),
        "high_cost_cases":
            (x["annual_medical_cost"] > 30000).sum(),
        "high_cost_pct":
            round(
                (x["annual_medical_cost"] > 30000).mean()*100,
                2
            )
    })
)

print(high_cost_region)
```

```python
# Calculate the Pearson correlation between age and annual medical cost
correlation = df["age"].corr(
    df["annual_medical_cost"]
)

print("Correlation =", round(correlation, 4))
```

```python
# Assign a risk tier (Low/Medium/High/Very High) based on smoking, BMI, and age, then summarize cost by tier
def risk_profile(row):

    if (
        row["smoker"] == 1 and
        row["bmi"] >= 30 and
        row["age"] >= 45
    ):
        return "Very High Risk"

    elif (
        row["smoker"] == 1 and
        row["bmi"] >= 30
    ):
        return "High Risk"

    elif (
        row["smoker"] == 1 or
        row["bmi"] >= 30
    ):
        return "Medium Risk"

    else:
        return "Low Risk"

df["risk_profile"] = df.apply(
    risk_profile,
    axis=1
)

risk_analysis = df.groupby(
    "risk_profile"
).agg(
    total_people=("risk_profile", "count"),
    avg_cost=("annual_medical_cost", "mean"),
    min_cost=("annual_medical_cost", "min"),
    max_cost=("annual_medical_cost", "max")
).round(2)

print(risk_analysis)
```

```python
# Divide the population into 10 equal cost groups (deciles) and analyze avg cost, age, and BMI per decile
df["cost_decile"] = pd.qcut(
    df["annual_medical_cost"],
    10,
    labels=False,
    duplicates="drop"
)

decile_analysis = df.groupby(
    "cost_decile"
).agg(
    total=("cost_decile", "count"),
    avg_cost=("annual_medical_cost", "mean"),
    avg_age=("age", "mean"),
    avg_bmi=("bmi", "mean")
).round(2)

print(decile_analysis)
```

```python
# Sort by cost descending and compute cumulative cost and each person's share of total spend
df_sorted = df.sort_values(
    "annual_medical_cost",
    ascending=False
)

df_sorted["cumulative_cost"] = (
    df_sorted["annual_medical_cost"].cumsum()
)

df_sorted["pct_of_total"] = (
    df_sorted["annual_medical_cost"]
    / df_sorted["annual_medical_cost"].sum()
    * 100
)

print(df_sorted.head())
```

```python
# Compute correlation of each key factor (age, BMI, income, etc.) with annual medical cost
factors = [
    "age",
    "bmi",
    "income",
    "risk_score",
    "chronic_count",
    "claims_count"
]

for col in factors:
    corr = df[col].corr(df["annual_medical_cost"])
    print(f"{col}: {round(corr,3)}")
```

```python
# Compare average cost, claims count, and chronic conditions between high-risk and low-risk groups
risk_analysis = df.groupby("is_high_risk").agg(
    avg_cost=("annual_medical_cost","mean"),
    avg_claims=("claims_count","mean"),
    avg_chronic=("chronic_count","mean")
).round(2)

print(risk_analysis)
```

```python
# Show average cost and average claims count grouped by number of chronic conditions
df.groupby("chronic_count").agg(
    avg_cost=("annual_medical_cost","mean"),
    avg_claims=("claims_count","mean")
).round(2)
```

```python
# Cross-analyze smoking status and chronic condition count to see their combined effect on cost
smoker_chronic = df.groupby(
    ["smoker","chronic_count"]
).agg(
    avg_cost=("annual_medical_cost","mean")
).round(2)

print(smoker_chronic)
```

```python
# Build a pivot table showing average cost broken down by BMI category (rows) and smoking status (columns)
pd.pivot_table(
    df,
    values="annual_medical_cost",
    index="bmi_category",
    columns="smoker",
    aggfunc="mean"
).round(2)
```

```python
# Compare average premium collected vs average cost paid out for high-risk vs low-risk members
premium_risk = df.groupby("is_high_risk").agg(
    avg_premium=("annual_premium","mean"),
    avg_cost=("annual_medical_cost","mean")
).round(2)

print(premium_risk)
```

```python
# Calculate each person's loss ratio (cost / premium), then average it by plan type
df["loss_ratio"] = (
    df["annual_medical_cost"] /
    df["annual_premium"]
)

df.groupby("plan_type").agg(
    avg_loss_ratio=("loss_ratio","mean")
).round(2)
```

```python
# Analyze how provider quality level affects average cost and average number of claims
provider_analysis = df.groupby(
    "provider_quality"
).agg(
    avg_cost=("annual_medical_cost","mean"),
    avg_claims=("claims_count","mean")
).round(2)

print(provider_analysis)
```

```python
# Compute correlation between each procedure type (imaging, surgery, etc.) and annual medical cost
procedures = [
    "proc_imaging_count",
    "proc_surgery_count",
    "proc_physio_count",
    "proc_consult_count",
    "proc_lab_count"
]

for col in procedures:
    print(
        col,
        round(df[col].corr(df["annual_medical_cost"]),3)
    )
```

```python
# Isolate the top 10% most expensive individuals (above 90th percentile) and show their stats
threshold = df["annual_medical_cost"].quantile(0.90)

top10 = df[
    df["annual_medical_cost"] >= threshold
]

top10.describe()
```

```python
# Count how many of the top 10% high-cost individuals are classified as high-risk
top10.groupby("is_high_risk").size()
```
